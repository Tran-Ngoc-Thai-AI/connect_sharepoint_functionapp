```python
import requests
```

```python
SEARCH_ENDPOINT = ""
SEARCH_API_KEY = ""
SEARCH_INDEX = ""
```

```python
def call_ai_search(search_request: dict) -> dict:
    """
    Input:
        search_request: JSON body gửi tới Azure AI Search.

    Output:
        JSON response nguyên bản từ Azure AI Search.
    """

    url = (
        f"{SEARCH_ENDPOINT.rstrip('/')}"
        f"/indexes('{SEARCH_INDEX}')"
        f"/docs/search"
    )

    response = requests.post(
        url,
        headers={
            "Content-Type": "application/json",
            "api-key": SEARCH_API_KEY,
        },
        params={
            "api-version": "2026-04-01",
        },
        json=search_request,
        timeout=30,
    )

    response.raise_for_status()

    return response.json()
```

```python
search_request = {
    "search": "Lê Đức Anh",
    "searchFields": "content",
    "top": 5
}
```

```python
result = call_ai_search(search_request)

result
```
